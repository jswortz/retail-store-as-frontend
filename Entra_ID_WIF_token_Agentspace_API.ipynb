```python
import requests
import json

def get_sts_token(subject_token=entra_token):
    url = "https://sts.googleapis.com/v1/token"

    data = {
        "audience": "//iam.googleapis.com/locations/global/workforcePools/entra-id-zken-personal/providers/entra-id",
        "grant_type": "urn:ietf:params:oauth:grant-type:token-exchange",
        "requested_token_type": "urn:ietf:params:oauth:token-type:access_token",
        "scope": "https://www.googleapis.com/auth/cloud-platform",
        "subject_token_type": "urn:ietf:params:oauth:token-type:saml2",
        "subject_token": subject_token, # Use the raw entra_token here, not the f-string version
        "options": "{\"userProject\":\"163236863044\"}"
    }

    response = requests.post(url, data=data)

    # To see the response status code
    print(f"Status Code: {response.status_code}")

    # To see the response content as JSON
    try:
        json_response = response.json()
        print("JSON Response:")
        display(json_response)
        return json_response['access_token']
    except json.JSONDecodeError:
        print("Response is not JSON:")
        print(response.text)

sts_token = get_sts_token()
```
